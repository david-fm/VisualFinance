```python
from visualfinance.data_providers import YahooFinanceDataProvider
from datetime import datetime
from dateutil.relativedelta import relativedelta
```

```python
Tickers = ['RHM.DE', 'HII', 'IDR.MC', 'GLNG', 'MC.PA', 'KER.PA', 'AMES.DE']
VentanaYears = 3
Today = datetime.now()
StartYear = Today - relativedelta(years=VentanaYears)

end_date = Today.date().strftime("%Y-%m-%d")
start_date = StartYear.date().strftime("%Y-%m-%d")

print(f"start_date: {start_date}")
print(f"end_date: {end_date}")
```

```python
YDP = YahooFinanceDataProvider()
data = YDP.get_prices(Tickers,start_date,end_date,interval="1wk",)
data.tail(5)
```
